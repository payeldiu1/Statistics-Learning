Exercise 1: Calculate the mean

Your bank recorded the following five transactions:
1000, 2000, 3000, 4000, 5000

Answer these questions:
1. What is the total transaction value?
2. How many transactions are there?
3. What is the mean transaction amount?

```python

import pandas as pd
txn = [1000, 2000, 3000, 4000, 5000]
df = pd.DataFrame(txn)
print('What is the total transaction value?')
print(df.sum())
print('How many transactions are there?')
print(df.count())
print('What is the mean transaction amount?')
print(df.mean())

print("\n")
import statistics
txn = [1000, 2000, 3000, 4000, 5000]
print('What is the total transaction value?')
print(sum(txn))
print('How many transactions are there?')
print(len(txn))
print('What is the mean transaction amount?')
print(statistics.mean(txn))


```


Exercise 2: Calculate the median

A bank records these customer account balances:
5000, 2000, 8000, 3000, 7000

Questions:
1. Arrange the values in ascending order.
2. Identify the middle value.
3. What is the median?


```python
import pandas as pd
txn = [5000, 2000, 8000, 3000, 7000]
df = pd.DataFrame({'amount': txn})
df_sort=df.sort_values('amount')
print(df_sort['amount'])
print('What is the median?')
print(df_sort['amount'].median())

print("\n")

import statistics
txn = [5000, 2000, 8000, 3000, 7000]
print("Sorted balances:", sorted(txn))

print("What is the median?",statistics.median(txn))
```
